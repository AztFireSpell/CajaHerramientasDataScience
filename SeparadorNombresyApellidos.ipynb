```python
import pandas as pd

def separar_nombres_apellidos(nombre_completo):
    # Si la celda está vacía, devolvemos espacios en blanco
    if pd.isna(nombre_completo) or str(nombre_completo).strip() == "":
        return pd.Series(["SIN INFORMACIÓN", "SIN INFORMACIÓN"])
        
    # 1. Limpieza inicial (ponemos la primera letra mayúscula de cada palabra)
    nombre = str(nombre_completo).strip().title()
    
    # 2. EL TRUCO: Unir las partículas compuestas temporalmente
    # NOTA: El orden es vital. Primero las largas ("De Los") y al final las cortas ("De")
    particulas = ["De Los ", "De Las ", "De La ", "Del ", "De ", "Y ", "San ", "Santa ", "Mc ", "Mac "]
    
    for p in particulas:
        # Reemplaza "De Los " por "De_Los_"
        nombre = nombre.replace(p, p.replace(" ", "_"))
        # Cubrimos también minúsculas por si el .title() falló en algún caso raro
        nombre = nombre.replace(p.lower(), p.lower().replace(" ", "_"))
        
    # 3. Separamos por espacios (ahora Python ve "María De_Los_Ángeles Pérez" como 3 palabras)
    partes = nombre.split()
    
    # 4. Aplicamos tu lógica analítica
    if len(partes) >= 3:
        # Los últimos dos elementos SIEMPRE serán los apellidos
        apellidos = " ".join(partes[-2:])
        # Todo lo que esté antes de los últimos dos, son los nombres
        nombres = " ".join(partes[:-2])
    elif len(partes) == 2:
        # Caso de "Alonso Pérez"
        nombres = partes[0]
        apellidos = partes[1]
    else:
        # Caso de que solo pusieron "Alonso"
        nombres = partes[0] if partes else ""
        apellidos = ""
        
    # 5. Limpieza final: Quitamos los guiones bajos para regresar al texto original
    nombres = nombres.replace("_", " ")
    apellidos = apellidos.replace("_", " ")
    
    # Retornamos dos valores para crear dos columnas nuevas
    return pd.Series([nombres, apellidos])

# --- CÓMO APLICARLO A TU DATAFRAME ---
# Supongamos que tu columna toda junta se llama 'Nombre_Completo'
# Esto creará las dos columnas mágicamente
# df[['Nombre_Extraido', 'Apellidos_Extraidos']] = df['Nombre_Completo'].apply(separar_nombres_apellidos)

# Puedes probar con algunos casos extremos:
casos_prueba = pd.DataFrame({'Nombre_Completo': [
    'Juan Carlos Perez Gomez',
    'Maria de los Angeles Macias del Prado',
    'Ana Maria de la Cruz de los Santos',
    'Jose Luis San Martin y Robles'
]})

casos_prueba[['Nombres', 'Apellidos']] = casos_prueba['Nombre_Completo'].apply(separar_nombres_apellidos)
casos_prueba
```
