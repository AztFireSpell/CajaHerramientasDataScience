```python
import pandas as pd
import os
```

```python
# --- CONFIGURACIÓN ---
archivo_origen = '#' # Cambia esto por la ruta de tu archivo original
columna_criterio = '#' # Columna que tomara el valor para separar en nombres de archivos
nombre_hoja = '#' #Por si tienes varias hojas aqui puedes ingresar Hoja1, Hoja2 etc
directorio_salida = '' # Carpeta donde se guardarán los Excel

#Se ingresan en formato de arreglo las columnas que quieres que se lleve el separador, ej. Nombre, Departamento, Correo, ETC
columnas_deseadas = []


# Crear el directorio de salida si no existe
if not os.path.exists(directorio_salida):
    os.makedirs(directorio_salida)

# --- PROCESAMIENTO ---
df = pd.read_csv(archivo_origen)
valores_unicos = df[columna_criterio].dropna().unique()

for valor in valores_unicos:
    nombre_archivo_seguro = str(valor).replace('/', '_').replace('\\', '_')
    ruta_guardado = os.path.join(directorio_salida, f"{nombre_archivo_seguro}.xlsx")
    
    # 1. Filtrar las filas correspondientes a este valor
    df_filtrado = df[df[columna_criterio] == valor]
    
    # 2. NUEVO: Filtrar solo las columnas que necesitas
    # Nota: Si una de las columnas deseadas no existe, Python lanzará un error aquí
    df_exportar = df_filtrado[columnas_deseadas]
    
    # 3. Guardar el resultado en Excel
    with pd.ExcelWriter(ruta_guardado, engine='openpyxl') as writer:
        df_exportar.to_excel(writer, sheet_name=nombre_hoja, index=False)
        
    print(f"✓ Archivo creado con columnas específicas: {ruta_guardado}")

print("\n¡Proceso de separación completado con éxito!")
```
